```python
import nlpaug.augmenter.audio as naa
import numpy as np
import librosa
import librosa.display as librosa_display
import matplotlib.pyplot as plt
import soundfile as sf
```

```python
list = "/home/sayed/espnet/egs/CHILD_ASR/LISTS/in_child_wav.list"

with open("/home/sayed/espnet/egs/CHILD_ASR/LISTS/in_child_wav.list","r") as org_file:
    org_items = [line.strip() for line in org_file if line.strip()]

with open("/home/sayed/espnet/egs/CHILD_ASR/LISTS/out_child_wav_crop.list","r") as crop_file:
    crop_items = [line.strip() for line in crop_file if line.strip()]
    
print(len(org_items),org_items[0],type(org_items[0]),crop_items[0],type(crop_items[0]))
```

```python
for i in range(len(org_items)):
    data, sr = librosa.load(org_items[i],sr=None)
    
    aug = naa.CropAug(sampling_rate=sr)
    augmented_data = aug.augment(data)

    augmented_data = np.array(augmented_data,dtype=np.float32)

    if len(augmented_data.shape) >1:
      augmented_data_new = augmented_data.flatten()

    sf.write(crop_items[i],augmented_data_new,sr)
```

```python
with open("/home/sayed/espnet/egs/CHILD_ASR/LISTS/out_child_wav_loud.list","r") as loud_file:
    loud_items = [line.strip() for line in loud_file if line.strip()]

print(len(loud_items),loud_items[0],type(loud_items[0]))
```

```python
for i in range(len(org_items)):
    data, sr = librosa.load(org_items[i],sr=None)
    
    aug2 = naa.LoudnessAug()
    augmented_data_loud = aug2.augment(data)

    augmented_data_loud = np.array(augmented_data_loud,dtype=np.float32)

    if len(augmented_data_loud.shape) >1:
      augmented_data_loud_new = augmented_data_loud.flatten()

    sf.write(loud_items[i],augmented_data_loud_new,sr)
```
